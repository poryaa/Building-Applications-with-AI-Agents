```python
from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage
from langgraph.graph import StateGraph, MessagesState

llm_model = ChatOllama(model="llama3.1:latest")

def call_model(state: MessagesState):
    response = llm_model.invoke(state["messages"])
    return {"messages": state["messages"] + [response]}

builder = StateGraph(MessagesState)
builder.add_node("model", call_model)
builder.set_entry_point("model")
graph = builder.compile()

# First turn
state = {"messages": [HumanMessage("Hi! I am porya.")]}
for chunk in graph.stream(state, stream_mode="values"):
    state = chunk
    state["messages"][-1].pretty_print()

# Second turn (with memory)
state["messages"].append(HumanMessage("What is my name?"))
for chunk in graph.stream(state, stream_mode="values"):
    state = chunk
    state["messages"][-1].pretty_print()
```
